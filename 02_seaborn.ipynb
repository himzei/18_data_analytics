```python

import seaborn as sns

titanic = sns.load_dataset("titanic")

titanic.head()
```

```python
titanic.info()
```

```python
# 회귀선이 있는 산점도 

import matplotlib.pyplot as plt 
import seaborn as sns

sns.set_style("darkgrid")

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.regplot(
    x="age", 
    y="fare",
    data=titanic, 
    ax=axes[0]
)

sns.regplot(
    x="age", 
    y="fare",
    data=titanic, 
    ax=axes[1],
    fit_reg=False
)

plt.show()
```

```python
titanic["survived"].value_counts()
```

```python
# 히스토그램/커널밀도함수

fig, axes = plt.subplot_mosaic(
    [["top_left", "top_right"],
     ["middle_left", "middle_right"],
     ["bottom", "bottom"]],
     figsize=(15, 6), 
     constrained_layout=True
)

sns.histplot(
    x="age",
    data=titanic, 
    ax=axes["top_left"]
)

sns.histplot(
    x="age",
    data=titanic, 
    
    hue="survived", 
    ax=axes["top_right"]
)

sns.histplot(
    x="age",
    data=titanic, 
    multiple="dodge",
    hue="survived", 
    ax=axes["middle_left"]
)

sns.histplot(
    x="age",
    data=titanic, 
    multiple="stack",
    hue="survived", 
    ax=axes["middle_right"]
)

sns.histplot(
    x="age",
    data=titanic, 
    multiple="fill",
    hue="survived", 
    bins=10, 
    ax=axes["bottom"]
)

plt.show()
```

```python
fig, axes = plt.subplot_mosaic(
    [
        ["top_left", "top_center", "right"],
        ["bottom_left", "bottom_center", "right"]
    ], 
    figsize=(15, 6), 
    constrained_layout=True)

sns.kdeplot(
    x="age", 
    data=titanic, 
    ax=axes["top_left"]
)

sns.kdeplot(
    x="age", 
    data=titanic, 
    hue="survived",
    ax=axes["bottom_left"]
)

sns.kdeplot(
    x="age", 
    data=titanic, 
    hue="survived",
    fill=True, 
    ax=axes["top_center"]
)

sns.kdeplot(
    x="age", 
    data=titanic, 
    hue="survived",
    multiple="stack",
    ax=axes["bottom_center"]
)

sns.kdeplot(
    x="age", 
    data=titanic, 
    hue="survived",
    multiple="fill",
    bw_adjust=1.0, 
    ax=axes["right"]
)

plt.show()
```

```python
sns.displot(
    titanic["fare"], 
    kind="hist"
)
plt.show()

sns.displot(
    titanic["fare"], 
    kind="kde"
)
plt.show()
```
