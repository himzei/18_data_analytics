```python
import pandas as pd
import matplotlib.pyplot as plt 

from matplotlib import font_manager, rc
font_path = "./data/malgun.ttf"
font_name = font_manager.FontProperties(fname=font_path).get_name()
rc("font", family=font_name)

plt.style.use("ggplot")
plt.rcParams["axes.unicode_minus"] = False
```

```python
df = pd.read_excel("./data/남북한발전전력량.xlsx")
df = df.loc[5:9]
df
```

```python
df.drop("전력량 (억㎾h)", axis="columns", inplace=True)
df.set_index("발전 전력별", inplace=True)
df
```

```python
df = df.T
df.head()
```

```python
# df.columns
for col in df.columns:
    df[col] = df[col].replace("-", "0")
    
print(df.head())
print(df.info())
```

```python
df = df.astype(float)
df.info()
```

```python
df.head()
```

```python
df = df.rename(columns={"합계":"총발전량"})
df.head()
```

```python
df["총발전량 - 1년"] = df["총발전량"].shift(1)
df.head()
```

```python
df["증감율"] = ((df["총발전량"] / df["총발전량 - 1년"]) - 1) * 100
df.head()
```

```python
ax1 = df[["수력", "화력"]].plot(
    kind="bar", 
    figsize=(20, 10), 
    width=0.7,
    stacked=True
)

ax2 = ax1.twinx()
ax2.plot(
    df.index, 
    df.증감율, 
    ls="--", 
    marker="o", 
    markersize=20, 
    color="green", 
    label="전년대비증감율"
)

ax1.set_ylim(0, 500)
ax2.set_ylim(-50, 30)
plt.show()
```

```python
# 히스토그램 
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("ggplot")

df = pd.read_csv("./data/auto-mpg.csv", header=None)
df.columns = [
    "mpg", "cylinders", "displacement", "horsepower", "weight", "acceleration", "model year", "origin", "name"
]

df["mpg"].plot(
    kind="hist", 
    bins=10, 
    color="coral", 
    figsize=(10, 5)
)


plt.show()


```

```python
df.head()
```

```python
df["origin"].value_counts()
```

```python
df[["mpg", "origin"]].plot(
    by=["origin"], 
    kind="hist", 
    figsize=(8, 10)
)
```

```python
df[["mpg", "origin", "cylinders"]].plot(
    by=["origin", "cylinders"], 
    kind="hist",
    figsize=(8, 30)
)
```

```python
# 산점도
plt.style.use("default")

df.plot(
    kind="scatter", 
    x="weight", 
    y="mpg", 
    c="coral", 
    s=10, 
    figsize=(10,5)
)

plt.show()
```

```python
plt.style.use("default")

cylinders_size = (df.cylinders / df.cylinders.max()) * 300

df.plot(
    kind="scatter", 
    x="weight", 
    y="mpg", 
    c="coral", 
    figsize=(10,5),
    s=cylinders_size, 
    alpha=0.3
)

plt.show()
```

```python
plt.style.use("default")

cylinders_size = (df.cylinders / df.cylinders.max()) * 300

df.plot(
    kind="scatter", 
    x="weight", 
    y="mpg", 
    marker="+",
    figsize=(10, 5),
    cmap="viridis", 
    c=cylinders_size, 
    s=50,
    alpha=0.3
)

plt.savefig("./data/scatter.png")
plt.show()
```
