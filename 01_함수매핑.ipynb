```python
# 시리즈원소에 함수 매핑
# Series객체.apply(매핑함수)

import seaborn as sns
import pandas as pd

titanic = sns.load_dataset("titanic")
df = titanic.loc[:, ["age", "fare"]]
df.head()
```

```python
def add_10(n): 
    return n + 10

def add_two_obj(a, b): 
    return a + b
```

```python
add_10(10)
add_two_obj(10, 10)
```

```python
df["age"].head()
```

```python
sr1 = df["age"].apply(add_10)
sr1.head()
```

```python
imsi = lambda n : n + 10
imsi(10)
```

```python
sr2 = df["age"].apply(lambda n : n + 10)
sr2.head()
```

```python
sr3 = df["age"].apply(add_two_obj, b=10)
sr3.head()
```

```python
sr4 = df["age"].apply(lambda a, b : a + b, b=10)
sr4
```
